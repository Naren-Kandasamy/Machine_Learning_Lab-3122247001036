```python
import numpy as np
import pandas as pd
import os
import kagglehub

from google.colab import userdata
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
os.environ["KAGGLE_API_TOKEN"] = userdata.get('KAGGLE_TOKEN')
download_path = kagglehub.competition_download('house-prices-advanced-regression-techniques')

train_path = os.path.join(download_path,'train.csv')
df_train = pd.read_csv(train_path)
print(df_train.columns)
```

```python
x = df_train['GrLivArea']
y = df_train['SalePrice']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2)

model = LinearRegression()
lin_reg = model.fit(x_train,y_train)
lin_reg_pred = model.predict(x_test)

mse = mean_squared_error(y_test,lin_reg_pred)
rmse = mse **0.5

print(f"Validation Mean Squared Error (MSE): {mse:.2f}")
print(f"Validation Root Mean Squared Error (RMSE): ${rmse:.2f}")
```
